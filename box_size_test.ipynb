```python


# Create dummy image for this example
img = np.zeros((200, 200), dtype=np.uint8)
cv2.rectangle(img, (67, 67), (133, 133), 255, -1)  # white square 66x66 in center

# Define original square size and center
square_size = 66
center = (img.shape[1] // 2, img.shape[0] // 2)  # (100, 100)

# Calculate coordinates of the original square
x1 = center[0] - square_size // 2
y1 = center[1] - square_size // 2
x2 = x1 + square_size
y2 = y1 + square_size

# Crop the white square
square = img[y1:y2, x1:x2]

# Resize to 10%
new_size = int(square_size * 1.1)
resized_square = cv2.resize(square, (new_size, new_size), interpolation=cv2.INTER_AREA)

# Create a blank image
output = np.zeros_like(img)

# Calculate new top-left corner to center the small square
new_x1 = center[0] - new_size // 2
new_y1 = center[1] - new_size // 2

# Paste resized square into output image
output[new_y1:new_y1+new_size, new_x1:new_x1+new_size] = resized_square

# Save or show result
#cv2.imwrite("resized_square_centered.png", output)
```

```python
plt.imshow(output)
```
