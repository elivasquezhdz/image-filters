```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
# Image dimensions
width, height = 400, 200
image = np.zeros((height, width, 3), dtype=np.uint8)

# Draw white circle on the left half
center_circle = (width // 4, height // 2)
radius = min(width, height) // 6
cv2.circle(image, center_circle, radius, (255, 255, 255), -1)

# Draw white square on the right half
square_size = min(width, height) // 3
top_left_square = (3 * width // 4 - square_size // 2, height // 2 - square_size // 2)
bottom_right_square = (top_left_square[0] + square_size, top_left_square[1] + square_size)
cv2.rectangle(image, top_left_square, bottom_right_square, (255, 255, 255), -1)

print()
```
