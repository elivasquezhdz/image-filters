```python
from PIL import Image
```

```python
input_path = "/Users/xixarron/Downloads/IMG_0548.JPG"
img = Image.open(input_path)
```

```python

def create_flipped_mosaic(img):
    # Load image
    
    # Create flipped versions
    h_flip = img.transpose(Image.FLIP_LEFT_RIGHT)   # horizontal flip
    v_flip = img.transpose(Image.FLIP_TOP_BOTTOM)  
    hv_flip = img.transpose(Image.FLIP_LEFT_RIGHT).transpose(Image.FLIP_TOP_BOTTOM)  # horizontal + vertical

    # Prepare output canvas (2x2 grid)
    w, h = img.size
    mosaic = Image.new("RGB", (w * 2, h * 2))

    # Place images
    mosaic.paste(img, (0, 0))            # top-left
    mosaic.paste(h_flip, (w, 0))         # top-right
    mosaic.paste(v_flip, (0, h))         # bottom-left
    mosaic.paste(hv_flip, (w, h))        # bottom-right

    return mosaic
```

```python
mosaic = create_flipped_mosaic(img)
```

```python
mosaic.save("mosaic4.png")
```
